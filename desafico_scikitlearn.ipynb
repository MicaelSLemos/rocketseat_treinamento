#Desafio Scikitlearn

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

#Bibliotecas do Scikit
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error
```

```python
dados_vendas = {
    'mes': ['Janeiro', 'Fevereiro', 'Março', 'Abril', 'Maio', 'Junho',  'Julho', 'Agosto', 'Setembro', 'Outubro', 'Novembro', 'Dezembro'],
    'vendas': [2000, 2200, 2300, 2500, 2600, 2700,2800, 2900, 3000, 3100, 3200, 3300]
    }
```

```python
df_vendas = pd.DataFrame.from_dict(dados_vendas)
df_vendas
```

```python
plt.plot(df_vendas['mes'], df_vendas['vendas'])
```

```python
df_vendas['m'] = df_vendas.index + 1
df_vendas.drop(columns=['mes'], inplace=True)
df_vendas = df_vendas.rename(columns={'m': "mes"})
```

```python
df_vendas
```

```python
X = df_vendas[['mes']]
y = df_vendas[['vendas']]
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7, random_state=51)
```

```python
#Criar modelo
modelo = LinearRegression().fit(X_train, y_train)
```

```python
#Testar
y_predict = modelo.predict(X_test)
```

```python
#Métricas
modelo.score(X_train, y_train)
```

```python
modelo.score(X_test, y_test)
```

```python
r2_score(y_test, y_predict)
```

```python
mean_absolute_error(y_test, y_predict)
```

```python
plt.scatter(X_test, y_test, color='g')
plt.plot(X_test, y_predict, color='k')
```

```python
print(f'Previsão Vendas Dezembro: {np.round(modelo.predict([[12]])[0][0],2)}')
```

```python
print('y=ax+b')
print(f'Coeficiente: {np.round(modelo.coef_[0][0],2)}; Intercepto: {np.round(modelo.intercept_[0],2)}')
```
